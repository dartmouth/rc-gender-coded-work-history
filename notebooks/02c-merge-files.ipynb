# Merge files with validated URLs


### Inputs: 
- `linkedin_company_urls_final.csv`

### Outputs:
...
  

### Purpose:

After the first run of getting and validating the URLs of all companies in the dataset, we changed our inclusion criteria and ended up with a larger number of companies. This added more URLs to the dataset, which require validation. Some of the URLs were validated manually. We therefore need to merge the manually validated URLs with the new currently unvalidated URLs.




```python
import pandas as pd
```

```python
human_validated = pd.read_csv("../data/derived/linkedin_company_urls_final.csv")
human_validated["Validated URL"] = human_validated["LinkedIn URL"]
```

```python
ai_validated = pd.read_csv("../data/derived/linkedin_company_urls_ai_validated.csv")
```

```python
ai_validated["LinkedIn URL"].isna().sum()
```

```python
all_urls = ai_validated.merge(human_validated, how="left")
```

```python
all_urls
```

```python
all_urls.loc[
    (all_urls["Is Valid"] == True) & (all_urls["Validated URL"].isna()), "Validated URL"
] = all_urls.loc[
    (all_urls["Is Valid"] == True) & (all_urls["Validated URL"].isna()), "LinkedIn URL"
]
```

```python
all_urls
```

```python
all_urls[all_urls["Validated URL"].isna()].to_excel(
    "../data/derived/flagged_linkedin_company_urls.xlsx", index=False
)

all_urls.dropna(subset=["Validated URL"]).to_csv(
    "../data/derived/linkedin_company_urls_precheck.csv", index=False
)
```
