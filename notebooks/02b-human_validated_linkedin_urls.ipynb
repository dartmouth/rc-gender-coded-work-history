# Add human-validated LinkedIn URLs

### Inputs: 
- `"../data/derived/flagged_linked_company_urls_jen_edits.xlsx"` (Only the manually corrected URLs flagged by AI)
- `"../data/derived/linkedin_company_missing_urls_added_simon_edits.xlsx"` (AI-validated URLs plus manually filled-in URLs missed by the Bing searches)


### Outputs:
- `"../data/derived/linkedin_company_urls_final.csv"`
  

### Purpose:

1. Merge human corrected data with automatically processed data

```python
import pandas as pd


human_corrected = pd.read_excel(
    "../data/derived/flagged_linked_company_urls_jen_edits.xlsx", na_values=["."]
)
```

Not all URls could be found. Determine the number of affected records:

```python
human_corrected[human_corrected["Human Correction"].isna()]["Affected Records"].sum()
```

A total of 79 records (counting each mention in admissions, interview and outcome data separately) are going to be affected by dropping the companies with missing LinkedIn URLs.

```python
human_corrected = human_corrected.dropna(subset="Human Correction")
```

Merge the corrected data with the AI-validated ones:

```python
ai_validated = pd.read_excel(
    "../data/derived/linkedin_company_missing_urls_added_simon_edits.xlsx"
)
```

```python
pd.concat(
    [
        ai_validated.loc[ai_validated["Is Valid"], ["Company Name", "LinkedIn URL"]],
        human_corrected[["Company Name", "LinkedIn URL"]],
    ]
).to_csv("../data/derived/linkedin_company_urls_final.csv", index=False)
```
