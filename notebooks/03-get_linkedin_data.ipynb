# Get LinkedIn data

### Inputs: 
- `"../data/derived/linkedin_company_urls_final.csv"`

### Outputs:

### Purpose:

1. Add company profile data from Bright Data dataset.

```python
import pandas as pd
```

```python
companies = pd.read_csv("../data/derived/linkedin_company_urls_final.csv")
# companies = pd.read_csv("../data/derived/linkedin_company_urls_precheck.csv")
```

```python
companies.columns
```

```python
companies["LinkedIn URL"]
```

```python
# Bright Data download requies the LinkedIn URL slugs
import re


def get_slug(url: str) -> str | None:
    match = re.search(r"company/([^/\s]+)(?:/|$)", url)
    if match:
        return match.group(1)
    return None


slugs = companies["LinkedIn URL"].apply(get_slug)
```

```python
slugs.to_csv("../data/derived/linkedin_company_slugs.csv", index=False, header=["id"])
```

Using the slugs, the [Bright Data LinkedIn dataset](https://brightdata.com/cp/datasets/browse/gd_l1vikfnt1wgvvqz95w?id=hl_0ba933c4) was filtered down to the available records:

```python
bd_data = pd.read_csv("../data/bd_20250725_201305_0.csv")
```

```python
companies = companies.set_index(slugs.rename("id"))
```

```python
bd_data = bd_data.set_index("id")
```

```python
companies = companies.merge(bd_data, on="id", how="left")
```

```python
companies = companies.rename(columns={"LinkedIn URL": "scraped_linkedin_url"})
companies.columns = ["_".join(c.lower().split()) for c in companies.columns]
companies
```

```python
companies.to_csv("../data/derived/company_profiles.csv")
```

## Determine the impact of missing companies

Let's determine how many samples we lose due to lack of company data.

```python
admissions_data = pd.read_csv("../data/derived/admissions.csv")

records = pd.Series(
    admissions_data["Job 1 Organization"].tolist(),
    name="Company",
)

record_count = records.value_counts().reset_index()
```

```python
missing_companies = companies[companies.industries.isna()]
```

```python
record_count[record_count["Company"].isin(missing_companies.company_name)].merge(
    missing_companies[["company_name", "scraped_linkedin_url"]],
    left_on="Company",
    right_on="company_name",
    how="left",
).drop(columns="company_name")
```
