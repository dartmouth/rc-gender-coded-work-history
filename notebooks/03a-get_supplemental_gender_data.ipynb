# Supplemental Gender Data

In previous work, manual industry classifications were made for a smaller dataset. We can leverage these categories in addition to the US Census categories.

In fact, these categories should be preferred because the corresponding percentages were researched to be representative of the actual industry.

This notebook generates a JSON file with those industry categories and the corresponding percentages.

```python
import pandas as pd
```

```python
df = pd.read_excel(
    "../data/supplemental/indus_gender_forsimon_cleaned.xlsx", na_values=["."]
)
```

```python
df.loc[df.indus == "Healthcare", ["indus_Simon", "indus_gen_percentage"]]
```

```python
df = df.dropna(subset=["indus_Simon", "indus_gen_percentage"])
```

```python
industries = df[["indus_Simon", "indus_gen_percentage"]].drop_duplicates()
industries = industries.rename(
    columns={"indus_Simon": "industry", "indus_gen_percentage": "women_pct"}
)
```

```python
industries.sort_values(by="industry")
```

```python
industries.to_csv("../data/derived/industry_gender_share.csv", index=False)
```
